```python
import pandas as pd

df = pd.read_csv("transformed_retail_sales.csv", parse_dates=["Order_Date"])
print(df.shape) 
```

```python
# Total revenue
total_revenue=df["Sales"].sum()
print("Total revenue:",total_revenue)
```

```python
# Total profit
total_profit=df["Profit"].sum()
print("Total Profit:",total_profit)
```

```python
# Total Orders
total_orders = len(df)      # every row is one order
print("Total Orders:", total_orders)
```

```python
# Total Customers
total_customers = df["Customer_ID"].nunique()
print("Total Customers:", total_customers)
```

```python
#Average Order Value
total_revenue = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = len(df)

aov = total_revenue / total_orders
print(f"Average Order Value:{aov:,.0f}")
```

```python
# Profit margin
total_revenue = df["Sales"].sum()
total_profit = df["Profit"].sum()
profit_margin=total_profit/total_revenue*100
print(f"Profit Margin: {profit_margin:.1f}%")
```

```python
# growth % in year compared 2025 to 2024
revenue_2024 = df[df["Year"] == 2024]["Sales"].sum()
revenue_2025 = df[df["Year"] == 2025]["Sales"].sum()
growth_pct=(revenue_2025-revenue_2024)/revenue_2024*100
print(f"Revenue Growth:{growth_pct:.1f}%")

profit_2024 = df[df["Year"] == 2024]["Profit"].sum()
profit_2025 = df[df["Year"] == 2025]["Profit"].sum()
print(f"Profit Growth: {(profit_2025 - profit_2024) / profit_2024 * 100:.1f}%")

orders_2024 = len(df[df["Year"] == 2024])
orders_2025 = len(df[df["Year"] == 2025])
print(f"Orders Growth: {(orders_2025 - orders_2024) / orders_2024 * 100:.1f}%")
```
